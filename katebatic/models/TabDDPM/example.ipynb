```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
abalone = fetch_ucirepo(id=1) 
  
# data (as pandas dataframes) 
X = abalone.data.features 
y = abalone.data.targets 

df = pd.concat([X,y], axis=1)
```

```python
X.drop(["Sex"], axis=1, inplace=True)
```

```python
X
```

```python
# example_usage.py

import pandas as pd
from tabddpm_wrapper import TabDDPM


# Initialize and train
model = TabDDPM(
    steps=1000,
    lr=0.001,
    weight_decay=1e-5,
    batch_size=4096,
    model_type='mlp',
    d_layers=[256, 256],
    dropout=0.0,
    num_classes=0,
    is_y_cond=False,
    normalization='quantile',
    device='cuda:0',
    seed=0
)

model.fit(X, y)

# Generate synthetic data
# X_syn, y_syn = model.generate(n_samples=100)
synthetic_df = model.generate_df(n_samples=100)

print(synthetic_df.head())
```
