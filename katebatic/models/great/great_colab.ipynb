```python
!git clone https://github.com/tabularis-ai/be_great.git
%pip install be-great kagglehub


from be_great import GReaT
from transformers import EarlyStoppingCallback
import pandas as pd
import kagglehub
import os

# 1) Download dataset from Kaggle
path = kagglehub.dataset_download("tejashvi14/tour-travels-customer-churn-prediction")
print("Path to dataset files:", path)

# Find the CSV file inside the downloaded dataset folder
files = os.listdir(path)
print("Files in dataset:", files)

# Assuming there's a main CSV file (adjust if different name)
csv_file = [f for f in files if f.endswith(".csv")][0]
data = pd.read_csv(os.path.join(path, csv_file))
print("Loaded dataset shape:", data.shape)

# 2) Create the model with stable args
model = GReaT(
    llm="distilgpt2",
    batch_size=32,
    epochs=100,                  # max cap; early stopping will cut it short
    fp16=True,
    dataloader_num_workers=4
)

# 3) Configure training strategies & early stopping (if supported)
if hasattr(model, "validation_split"):
    model.validation_split = 0.15
elif hasattr(model, "eval_dataset_size"):
    model.eval_dataset_size = 0.15

if hasattr(model, "evaluation_strategy"):
    model.evaluation_strategy = "epoch"
if hasattr(model, "save_strategy"):
    model.save_strategy = "epoch"
if hasattr(model, "logging_strategy"):
    model.logging_strategy = "epoch"

if hasattr(model, "load_best_model_at_end"):
    model.load_best_model_at_end = True
if hasattr(model, "metric_for_best_model"):
    model.metric_for_best_model = "eval_loss"
if hasattr(model, "greater_is_better"):
    model.greater_is_better = False

if hasattr(model, "early_stopping_patience"):
    model.early_stopping_patience = 3

# Add EarlyStoppingCallback if callbacks attribute exists
if hasattr(model, "callbacks"):
    if model.callbacks is None:
        model.callbacks = [EarlyStoppingCallback(early_stopping_patience=3)]
    else:
        model.callbacks.append(EarlyStoppingCallback(early_stopping_patience=3))

# 4) Train the model
model.fit(data)

# 5) Generate synthetic samples and save
synthetic_data = model.sample(n_samples=100)
print(synthetic_data.head())
synthetic_data.to_csv("synthetic_customertravel.csv", index=False)
print("Saved to synthetic_customertravel.csv")
```

X_train- original data features, y-train-original data target, x-test- synthetic

```python
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, classification_report,
    mean_squared_error, mean_absolute_error, r2_score
)
import numpy as np

def evaluate_models(X_train, X_test, y_train, y_test, task="classification"):
    assert task in ["classification", "regression"], "Task must be 'classification' or 'regression'"

    if task == "classification":
        models = {
            "Logistic Regression": LogisticRegression(max_iter=1000),
            "Decision Tree": DecisionTreeClassifier(),
            "Random Forest": RandomForestClassifier()
        }
    else:  # regression
        models = {
            "Linear Regression": LinearRegression(),
            "Decision Tree": DecisionTreeRegressor(),
            "Random Forest": RandomForestRegressor()
        }

    for name, model in models.items():
        print(f"\n=== {name} ===")
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        if task == "classification":
            print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
            print(f"Precision: {precision_score(y_test, y_pred, average='weighted'):.4f}")
            print(f"Recall: {recall_score(y_test, y_pred, average='weighted'):.4f}")
            print(f"F1 Score: {f1_score(y_test, y_pred, average='weighted'):.4f}")
            print("Classification Report:")
            print(classification_report(y_test, y_pred))
        else:  # regression
            print(f"Mean Squared Error (MSE): {mean_squared_error(y_test, y_pred):.4f}")
            print(f"Root Mean Squared Error (RMSE): {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
            print(f"Mean Absolute Error (MAE): {mean_absolute_error(y_test, y_pred):.4f}")
            print(f"R² Score: {r2_score(y_test, y_pred):.4f}")
```

```python
X_train = data.drop(["Target"], axis=1)
y_train = data["Target"]
X_test = synthetic_data.drop(["Target"], axis=1)
y_test = synthetic_data["Target"]
from sklearn.preprocessing import LabelEncoder

# Identify categorical columns
categorical_cols = X_train.select_dtypes(include=['object', 'category']).columns

# Initialize a dictionary to store encoders for each column
encoders = {}

# Apply LabelEncoder to each categorical column
for col in categorical_cols:
    le = LabelEncoder()
    X_train[col] = le.fit_transform(X_train[col])
    X_test[col] = le.transform(X_test[col])  # Use same encoder for consistency
    encoders[col] = le  # Save encoder in case needed later


```

```python
evaluate_models(X_train, X_test, y_train, y_test, task="classification")
```
