The pipeline does not support data preprocessing. All csv file needs to be cleaned, no NaN values before fitting.

```python
from original_pategans import *
```

```python
pategans = PateGans(
    epsilon=5,     
    delta=1e-5,
    num_teachers=10
)

synthetic_data = pategans.fit_generate("raw_data/train_epileptic_seizure.csv", n_samples=9200)
display("Synthetic data shape:", synthetic_data.shape)
display(synthetic_data.describe())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

X_synthetic = synthetic_data.drop('y', axis=1)
y_synthetic = synthetic_data['y']

test_data = pd.read_csv("raw_data/test_epileptic_seizure.csv")
X_test = test_data.drop('y', axis=1)
y_test = test_data['y']

# Train Random Forest on synthetic data
rf = RandomForestClassifier(random_state=42)
rf.fit(X_synthetic, y_synthetic)

# Get predicted probabilities for the positive class (class 1: seizure)
y_prob = rf.predict_proba(X_test)[:, 1]

# Calculate AUROC
auroc = roc_auc_score(y_test, y_prob)
print(f"AUROC: {auroc:.4f}")
```
