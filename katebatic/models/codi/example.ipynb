# Create synthetic dataset from Codi

```python
from codi import *
```

```python
synthetic_data = codi(
    csv_path='raw_data/heart_train.csv',
    test_split=0.2,
    total_epochs_both=1000,
    training_batch_size=512,
    num_samples=814,
    verbose=False
)
```

```python
display(synthetic_data.head())
display(synthetic_data.describe())
display(synthetic_data.shape)
```

```python
import pandas as pd

# Convert all columns to float
for col in synthetic_data.columns:
    synthetic_data[col] = pd.to_numeric(synthetic_data[col], errors='coerce').astype(float)

synthetic_data.info()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
# Prepare the data
X_synthetic = synthetic_data.drop('target', axis=1) 
y_synthetic = synthetic_data['target']

test_data = pd.read_csv("raw_data/heart_test.csv")
X_test = test_data.drop('target', axis=1)
y_test = test_data['target']

# Train Random Forest on synthetic data
rf = RandomForestClassifier(random_state=42)
rf.fit(X_synthetic, y_synthetic)

# Make predictions on test data
y_pred = rf.predict(X_test)
y_pred = y_pred.astype(int)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
