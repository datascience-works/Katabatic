```python
import numpy as np
import pandas as pd
from pathlib import Path

real_path = Path("processed/adult.binary.matrix")
syn_path  = Path("runs/adult_medgan/adult.synthetic.npy")

# real was saved as np.save in your pipeline (we used np.load earlier successfully)
X_real = np.load(real_path, allow_pickle=True).astype(np.float32)
X_syn  = np.load(syn_path).astype(np.float32)

print("Real:", X_real.shape, "Synthetic:", X_syn.shape)
```

```python
def top_abs_diff(means_a, means_b, k=10):
    diff = np.abs(means_a - means_b)
    idx = np.argsort(-diff)[:k]
    return pd.DataFrame({
        "col": idx,
        "mean_real": means_a[idx],
        "mean_syn": means_b[idx],
        "abs_diff": diff[idx],
    })

real_means = X_real.mean(axis=0)
syn_means  = X_syn.mean(axis=0)
real_stds  = X_real.std(axis=0)
syn_stds   = X_syn.std(axis=0)

print("Mean difference (L1):", float(np.abs(real_means - syn_means).mean()))
print("Std  difference (L1):", float(np.abs(real_stds  - syn_stds ).mean()))

top10 = top_abs_diff(real_means, syn_means, k=10)
top10
```

```python

import os
from pathlib import Path

os.environ["OMP_NUM_THREADS"] = "1"
os.environ["MKL_NUM_THREADS"] = "1"


Path("runs/joblib_tmp").mkdir(parents=True, exist_ok=True)
os.environ["JOBLIB_TEMP_FOLDER"] = str(Path("runs/joblib_tmp").absolute())

# ==== 1) IMPORTS ====
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier


try:
    from xgboost import XGBClassifier
    HAVE_XGB = True
except Exception:
    HAVE_XGB = False

# ==== 2) PATHS & LABEL COLUMNS ====
REAL_PATH = Path("processed/adult.binary.matrix")              # your real matrix (.npy)
SYN_PATH  = Path("runs/adult_medgan/adult.synthetic.npy")      # synthetic produced by medGAN

# >>>>>> Adjust these two if needed <<<<<<
LABEL_COL_REAL = 6     # label column index in the REAL matrix (from your earlier print)
LABEL_COL_SYN  = 50    # label column index in the SYNTHETIC matrix (from your earlier print)

# ==== 3) LOAD DATA ====
# Real (your file loads via np.load(..., allow_pickle=True))
Xr_all = np.load(REAL_PATH, allow_pickle=True)
# Synthetic (is a standard .npy produced by our sampling step)
Xs_all = np.load(SYN_PATH)

print("Real shape:", Xr_all.shape)
print("Synthetic shape:", Xs_all.shape)

# ==== 4) SPLIT INTO (X, y) AND CAST TYPES ====
def split_xy(arr, label_col):
    y = arr[:, label_col]
    X = np.delete(arr, label_col, axis=1)
    return X, y

X_real_all, y_real_all = split_xy(Xr_all, LABEL_COL_REAL)
X_syn_all,  y_syn_all  = split_xy(Xs_all, LABEL_COL_SYN)

# Make dtypes explicit (more stable & lighter)
X_real_all = np.asarray(X_real_all, dtype=np.float32)
X_syn_all  = np.asarray(X_syn_all,  dtype=np.float32)
y_real_all = np.asarray(y_real_all, dtype=np.int32)
y_syn_all  = np.asarray(y_syn_all,  dtype=np.int32)

# Train/Test split on REAL for evaluation (we test on real!)
X_real_train, X_real_test, y_real_train, y_real_test = train_test_split(
    X_real_all, y_real_all, test_size=0.2, stratify=y_real_all, random_state=42
)

print(f"Real (train/test): {X_real_train.shape} / {X_real_test.shape}")
print(f"Synthetic (all):   {X_syn_all.shape}")

# ==== 5) DEFINE MODELS (single-threaded for notebook stability) ====
models = {
    "LogisticRegression": Pipeline([
        ("scaler", StandardScaler(with_mean=False)),
        ("clf", LogisticRegression(solver="lbfgs", max_iter=400))
    ]),
    "MLP": Pipeline([
        ("scaler", StandardScaler(with_mean=False)),
        ("clf", MLPClassifier(hidden_layer_sizes=(128,), max_iter=80, random_state=42))
    ]),
    "RandomForest": RandomForestClassifier(
        n_estimators=200, n_jobs=1, random_state=42
    ),
}
if HAVE_XGB:
    models["XGBoost"] = XGBClassifier(
        n_estimators=300, max_depth=6, learning_rate=0.1,
        subsample=0.9, colsample_bytree=0.9,
        n_jobs=1, tree_method="hist",
        eval_metric="logloss", random_state=42
    )

# Optional: if you want a quicker debug run, cap the synthetic training size
MAX_SYN = None   # e.g., set to 20000 for a quick pass; leave None for full
if MAX_SYN is not None and len(X_syn_all) > MAX_SYN:
    X_syn_use = X_syn_all[:MAX_SYN]
    y_syn_use = y_syn_all[:MAX_SYN]
else:
    X_syn_use = X_syn_all
    y_syn_use = y_syn_all

# ==== 6) TSTR: Train on SYN, Test on REAL ====
rows = []
for name, model in models.items():
    try:
        # Fit on synthetic
        model.fit(X_syn_use, y_syn_use)

        # Predict proba if available, else use decision_function and min-max to [0,1]
        if hasattr(model, "predict_proba"):
            y_proba = model.predict_proba(X_real_test)[:, 1]
        elif hasattr(model, "decision_function"):
            s = model.decision_function(X_real_test)
            smin, smax = s.min(), s.max()
            y_proba = (s - smin) / (smax - smin + 1e-12)
        else:
            y_proba = None

        y_pred = model.predict(X_real_test)

        acc = accuracy_score(y_real_test, y_pred)
        f1  = f1_score(y_real_test, y_pred)
        auc = roc_auc_score(y_real_test, y_proba) if y_proba is not None else np.nan

        rows.append({"Model": name, "Accuracy": acc, "F1": f1, "AUC": auc})
        print(f"{name:>16}  Acc={acc:.4f}  F1={f1:.4f}  AUC={auc if not np.isnan(auc) else float('nan'):.4f}")

    except Exception as e:
        rows.append({"Model": name, "Accuracy": np.nan, "F1": np.nan, "AUC": np.nan})
        print(f"[WARN] {name} failed: {e}")

results = pd.DataFrame(rows).sort_values("AUC", ascending=False).reset_index(drop=True)
results
```
