
# ARF Data Generation


```python

from katabatic.models.arf.models import ARFModel
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from utils import discretize_preprocess
```

```python

dataset_path = "raw_data/shuttle.csv"
output_path = "discretized_data/shuttle.csv"
discretize_preprocess(dataset_path, output_path)
```

```python

datasets = ['car', 'adult', 'magic', 'nursery', 'shuttle']

for dataset in datasets:
    input_csv = f'discretized_data/{dataset}.csv'
    output_dir = f'sample_data/{dataset}'
    
    pipeline = TrainTestSplitPipeline(model=ARFModel)
    pipeline.run(
        input_csv=input_csv,
        output_dir=output_dir,
        synthetic_dir=f'synthetic/{dataset}/arf',
        real_test_dir=f'sample_data/{dataset}'
    )
```
