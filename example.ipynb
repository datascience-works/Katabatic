```python
input_csv = 'encoded_data/car.csv'

# Ensure all imports are available in this cell
from pathlib import Path
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from katabatic.models.tabsyn.models import TabSyn
from katabatic.artifacts import LocalArtifactStore

import torch
print('torch version:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())

artifact_store = LocalArtifactStore(Path('artifacts'))

try:
    pipeline = TrainTestSplitPipeline(model=TabSyn)
    result = pipeline.run(
        input_csv=input_csv,
        dataset_name='car',
        artifact_store=artifact_store,
        model_name='tabsyn',
        extra_source_dir='sample_data/car',
    )
    print(result)
except Exception as e:
    import traceback
    print('Exception:', e)
    traceback.print_exc()
```
