```python
# example_usage.py
import pandas as pd
from katebatic.models.TabDDPM import tabddpm

# from sklearn.linear_model import LinearRegression
```

```python
import os
os.getcwd()
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
abalone = fetch_ucirepo(id=1) 
  
# data (as pandas dataframes) 
X = abalone.data.features 
y = abalone.data.targets 

df = pd.concat([X,y], axis=1)
```

```python
X
```

```python
X.drop(["Sex"], axis=1, inplace=True)
```

```python
X
```

```python
# example_usage.py

# Initialize and train
model = tabddpm.TabDDPM(
    steps=1000,
    lr=0.001,
    weight_decay=1e-5,
    batch_size=4096, 
    model_type='mlp', 
    d_layers=[256, 256],
    dropout=0.0,
    num_classes=0,
    is_y_cond=False,
    normalization='quantile',
    device='cpu',
    seed=1
)

model.fit(X, y)

# Generate synthetic data
# X_syn, y_syn = model.generate(n_samples=100)
synthetic_df = model.generate_df(n_samples=150)

print(synthetic_df.head())
```

# TSTR - For Classification

```python
import pandas
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X, y)
```

```python
X_syn = synthetic_df.drop(["target"], axis=1)
y_syn = synthetic_df['target']
```

```python
predictions = model.predict(X_syn)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
cf = confusion_matrix(y_syn, predictions)
print(cf)
```

```python
accuracy_score(y_syn, predictions)
```
