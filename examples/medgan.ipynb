```python
# Imports
import sys
from pathlib import Path

# Resolve project root and ensure it's on sys.path
ROOT = Path.cwd().resolve()
for _ in range(5):
    if (ROOT / "pyproject.toml").exists() or (ROOT / "raw_data").exists():
        break
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
# Preprocess data
from pathlib import Path

from katabatic.models.medgan.models import MEDGAN
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from katabatic.utils.preprocess import preprocess_dataset #, preprocess_tabular

dataset_path = ROOT / "raw_data" / "car.csv"
output_path = ROOT / "preprocessed_data" / "car.csv"
output_path.parent.mkdir(parents=True, exist_ok=True)

preprocess_dataset(str(dataset_path), str(output_path))
```

```python
from katabatic.artifacts import LocalArtifactStore

input_csv = str(ROOT / "preprocessed_data" / "car.csv")

artifact_store = LocalArtifactStore(ROOT / "artifacts")

pipeline = TrainTestSplitPipeline(
    model=MEDGAN(latent_dim=64)
)

result = pipeline.run(
    input_csv=input_csv,
    dataset_name="car",
    artifact_store=artifact_store,
    model_name="medgan",
)

print(result)
```
